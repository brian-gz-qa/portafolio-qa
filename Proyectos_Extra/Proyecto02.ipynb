```python
import pyautogui
import pyperclip
import webbrowser
import time  
```

```python
webbrowser.open("https://mail.google.com/mail/u/0/#inbox")

destinatario = "micorreo@gmail.com"

asunto = "Análisis de acciones últimos 6 meses"

mensaje = f"""

Estimado/a,

Acá le envío el análisis de los últimos 6 meses de la acción que solicitó.

Cotización máxima: {maxima} 
Cotización mínima: {minima}
Cotización media: {valor_medio}

Espero que esta información le sea útil.
Quedo a su disposición para cualquier consulta adicional.

saludos cordiales


"""

time.sleep(5)  # Esperar a que se abra el navegador


pyautogui.PAUSE = 3  # Pausa de 1 segundo entre acciones de PyAutoGUI

pyautogui.click(187, 310)  # Hacer clic en la posición deseada (x=187, y=310)

pyperclip.copy(destinatario)  # Copiar el destinatario al portapapeles
pyautogui.hotkey("ctrl", "v")  # Pegar el destinatario

pyautogui.press("tab")  # Presionar la tecla Tab para ir al campo de asunto
pyperclip.copy(asunto)  # Copiar el asunto al portapapeles
pyautogui.hotkey("ctrl", "v")  # Pegar el asunto

pyautogui.press("tab")  # Presionar la tecla Tab para ir al campo de mensaje
pyperclip.copy(mensaje)  # Copiar el mensaje al portapapeles
pyautogui.hotkey("ctrl", "v")  # Pegar el mensaje

pyautogui.click(2749, 2018)  # Hacer clic en el botón de enviar (x=2749, y=2018)
pyautogui.hotkey("ctrl", "w")  # Cerrar la pestaña del navegador
```
