```python
import yfinance as yf 
```

```python
# Cargar datos y extraer precios de cierre
datos = yf.download(["AAPL","NVDA", "MSFT"], period="1y")
precios = datos["Close"]
```

```python
precios.tail()
```

```python
precios.describe()
```

## Gráficos interactivos

```python
%pip install plotly
import plotly.express as px
```

```python
%pip install nbformat ipython


px.line(precios,
        title="Cotización Historica: Apple (ultimos 12 meses)", 
        labels={"value": "Precio (USD)", "Date": "Fecha"},
        template = "plotly_dark")
```

```python
px.bar(
    precios.mean(),
    title="Cotización promedio: Apple (últimos 12 meses)",
    labels={"value": "Precio (USD)", "index": "Ticker"},
    template="plotly_dark"
)

```

## Acceder a las filas

```python
precios.head()
```

```python
precios.values[-1]
```

```python
inversion = (precios / precios.values[0]) * 1000

px.line(inversion,
        title="Evolución de una inversión de $1000: Apple (últimos) 12 meses)",
        labels={"value": "Valor de la inversión (USD)", "Date": "Fecha"},
        template="plotly_dark"
)
```

## Cómo guardar gráficas

```python
fig = px.line(inversion,
        title="Evolución de una inversión de $1000: Apple (últimos) 12 meses)",
        labels={"value": "Valor de la inversión (USD)", "Date": "Fecha"}, 
        template="plotly_dark"
)
```

```python
fig.write_html("inversion de 1000 USD.html")
```

## 5. Informe con IA

```python
from google import genai as gemini 
```

```python
from google import genai

# 1. Crear el cliente
cliente = genai.Client(api_key="TU_API_KEY_DE_GEMINI")

# 2. Definir el prompt
prompt = f"""haz un analisis del ultimo año de las empresas apple, nvidia y microsoft con vase en noticias donde hubo oportunidades de inversion para multiplicar el capital invertido, y que se pueda ver en la siguiente tabla de precios de cierre de las acciones de estas empresas:
{precios}"""

# 3. Generar contenido (models en plural y nombre de modelo válido)
response = cliente.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

# 4. Imprimir respuesta
print(response.text)
```
